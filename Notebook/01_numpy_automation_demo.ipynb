## Notebook 1: NumPy Automation Demo

Notebook ini mendemonstrasikan berbagai fitur otomasi menggunakan NumPy
untuk data processing, feature engineering, dan analisis statistik.

```python
import sys
import os
import numpy as np
import matplotlib.pyplot as plt

# Add src to path
sys.path.insert(0, os.path.join(os.getcwd(), '..'))
from src.numpy_automation import NumPyAutomation

print("NumPy version:", np.__version__)
print("Ready!")
```

## 1. Initialize & Generate Data

```python
auto = NumPyAutomation(random_seed=42)

# Generate synthetic data
X, y = auto.generate_synthetic_data(
    n_samples=2000,
    n_features=10,
    task_type='classification'
)

print(f"Features shape: {X.shape}")
print(f"Labels shape: {y.shape}")
print(f"Class distribution: {np.bincount(y)}")
```

## 2. Data Preprocessing

```python
# Normalize data
X_norm, params = auto.normalize(X, method='zscore')

# Split data
splits = auto.split_data(X_norm, y, test_ratio=0.2, val_ratio=0.1)

print(f"Train: {splits['X_train'].shape}")
print(f"Val:   {splits['X_val'].shape}")
print(f"Test:  {splits['X_test'].shape}")
```

## 3. Feature Engineering

```python
# Polynomial features
X_poly = auto.create_polynomial_features(splits['X_train'], degree=2)
print(f"Polynomial features: {splits['X_train'].shape} -> {X_poly.shape}")

# Statistical features
X_stats = auto.extract_statistical_features(splits['X_train'])
print(f"Statistical features: {X_stats.shape}")
```

## 4. PCA Visualization

```python
X_pca, eigenvalues, eigenvectors = auto.compute_pca_manual(splits['X_train'], n_components=2)

plt.figure(figsize=(10, 6))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=splits['y_train'],
                       cmap='viridis', alpha=0.6, s=20)
plt.colorbar(scatter, label='Class')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('PCA Projection (2 Components)')
plt.tight_layout()
plt.show()
```

## 5. Statistics & Correlation

```python
stats = auto.compute_statistics(splits['X_train'])
corr = auto.compute_correlation_matrix(splits['X_train'])

# Plot correlation heatmap
plt.figure(figsize=(10, 8))
plt.imshow(corr, cmap='RdBu_r', vmin=-1, vmax=1)
plt.colorbar(label='Correlation')
plt.title('Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```

## 6. Outlier Detection

```python
outlier_mask, outlier_idx = auto.detect_outliers_iqr(splits['X_train'], threshold=1.5)

# Visualize outliers in PCA space
plt.figure(figsize=(10, 6))
plt.scatter(X_pca[~outlier_mask, 0], X_pca[~outlier_mask, 1],
            c='blue', alpha=0.5, s=20, label='Normal')
plt.scatter(X_pca[outlier_mask, 0], X_pca[outlier_mask, 1],
            c='red', alpha=0.8, s=40, label='Outlier')
plt.legend(loc='best')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('Outlier Detection (IQR Method)')
plt.tight_layout()
plt.show()
```
